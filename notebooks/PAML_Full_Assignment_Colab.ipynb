## Task 2 – K-Means Clustering

The dataset is stored as a NumPy array representation inside a text file. We read it as raw text and safely convert it into a numerical NumPy array.

```
import ast
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```
with open('Task2_Data_Clustering.txt', 'r') as f:
    content = f.read()

content = content.replace('array(', '').rstrip(')')
data = np.array(ast.literal_eval(content), dtype=float)

print('Data shape:', data.shape)
print('Data type:', data.dtype)
```

```
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)
```

```
kmeans = KMeans(n_clusters=3, random_state=42)
clusters = kmeans.fit_predict(scaled_data)
```

```
plt.scatter(scaled_data[:, 0], scaled_data[:, 1], c=clusters)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('K-Means Clustering Result')
plt.show()
```

## Task 3 – Regression Analysis

A second-degree polynomial regression model is applied to analyse portfolio returns as a function of risk.

```
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
```

```
df = pd.read_csv('Task3_Data_Process.csv')
X = df[['x']]
y = df['y']
```

```
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)

y_pred = model.predict(X_poly)
```

```
print('R2 Score:', r2_score(y, y_pred))
print('Mean Squared Error:', mean_squared_error(y, y_pred))
```

```
sorted_idx = X.values.flatten().argsort()
X_sorted = X.values.flatten()[sorted_idx].reshape(-1, 1)
X_poly_sorted = poly.transform(X_sorted)
y_pred_sorted = model.predict(X_poly_sorted)

plt.scatter(X, y)
plt.plot(X_sorted, y_pred_sorted)
plt.xlabel('Risk Factor (x)')
plt.ylabel('Portfolio Return (y)')
plt.title('Polynomial Regression Model')
plt.show()
```

## Conclusion

- K-Means clustering identified three distinct clusters.
- Polynomial regression revealed a negative non-linear relationship between risk and returns.
- Results are consistent with expectations for financial data.
